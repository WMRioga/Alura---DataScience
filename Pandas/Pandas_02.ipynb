# Banco de datos SQL

```python
import pandas as pd
import sqlalchemy as sq
from sqlalchemy import create_engine, MetaData, Table, inspect, text
```

```python
# 1. Creamos el engine
# engine_1 = create_engine('postgresql://postgres:postgres@localhost:5432/postgres')
engine_2 = create_engine('sqlite:///:memory:') # En memoria
```

```python
# Para poder guardar un sql, la informacion debe ser tipo dataframe.
# df_emisiones_C02.to_sql('emisiones_C02', engine_1, index=False) 
# Para guardar un csv en sql, se debe primero pasar a dataframe
datos_csv = pd.read_csv('./1980-aula1/data/superstore_data.csv')
datos_csv
```

```python
datos_csv.to_sql('superstore_df', engine_2, index=False)    
```

```python
inspector = inspect(engine_2)
inspector.get_table_names()
```

```python
query = text('SELECT * FROM superstore_df WHERE Education = "Graduation" AND Year_Birth > 1970')
df = pd.read_sql(query, engine_2)
df
```

```python
df.to_sql('superstore_df_2_grad', engine_2, index=False)
```

```python
inspector = inspect(engine_2)
inspector.get_table_names()
```

```python
pd.read_sql('superstore_df_2_grad', engine_2, columns=['Id', 'Year_Birth', 'Education', 'Marital_Status', 'Income']) 
```

```python
pd.read_sql('superstore_df_2_grad', engine_2, columns=[0, 1, 2, 3, 4]) 
```

```python
query = text('SELECT * FROM superstore_df WHERE Education = "Graduation" AND Year_Birth > 1970')
df = pd.read_sql(query, engine_2)    
df  
```

```python
# Borrar registros.
from sqlalchemy.exc import SQLAlchemyError

query = text('DELETE FROM superstore_df WHERE Id = 0 or Id = 1')
try:
    r_set = engine_2.connect().execute(query)
except SQLAlchemyError as e:
    print(e)
else:
    print("Registros borrados exitosamente.", r_set.rowcount)
```

```python
# Cambiar un registro.
from sqlalchemy.exc import SQLAlchemyError

query = text('UPDATE superstore_df SET Education = "Master" WHERE Id = 22')
try:
    r_set = engine_2.connect().execute(query)
except SQLAlchemyError as e:
    print(e)
else:
    print("Registros actualizados exitosamente.", r_set.rowcount)
```

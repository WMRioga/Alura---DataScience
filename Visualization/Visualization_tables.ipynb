# Mision 1: Equipo Ventas
>
> Se necesita construir una visualizacion para indicar __los 10 principales clientes por total de ventas__.
--- 

```python
clientes = df_01.groupby('nombre_cliente')['ventas'].sum().sort_values(ascending=False).copy().round(2).head(10)     ## sum().nlargest(10)
clientes = clientes.reset_index()
clientes
```

```python
clientes.columns = ['Clientes', 'Ventas']
clientes['Ranking'] = clientes['Ventas'].rank(ascending=False).astype(int)
clientes.set_index('Ranking', inplace=True)
clientes
```

```python
estilo_01 = clientes.style
estilo_01
```

```python
estilo_01.format({'Ventas': '${:,.2f}'})
estilo_01
```

# Mision 3: Equipo de Logistica
> 
> Se necesita construir una visualizacion que nos permita entender __la distribucion de pedidos por region en Argentina__.
---

```python
import numpy as np

region = pd.DataFrame(df_01['region']).value_counts().to_frame().reset_index()
region.columns = ['Region', 'Num pedidos']
region.set_index('Region', inplace=True)
region
```

```python
# porcentaje = region['Num pedidos'] / region['Num pedidos'].sum() * 100
# region['Porcentaje'] = porcentaje.round(2)
# region
```

```python
porcentaje = region['Num pedidos'].to_numpy()
porcentaje = porcentaje / np.sum(porcentaje) * 100
region['Porcentaje'] = porcentaje.round(2)
region
```

```python
estilo_03 = region.style
estilo_03
```

```python
encabezado_03 = {
    'selector': 'th',
    'props': [
        ('background-color', 'white'),
        ('color', 'black'),
        ('font-family', 'Arial'),
        ('font-size', '16px'),
        ('font-weight', 'bold'),
        ('text-align', 'right'),
        ('text-transform', 'capitalize'),
        ('padding', '5px')
    ]
}
celdas_03 = {
    'selector': 'td',
    'props': [
        ('background-color', 'white')
    ]
}
```

```python
estilo_03.set_table_styles([encabezado_03, celdas_03])
```

```python
estilo_03.format({'Porcentaje': '{:,.2f}%'})\
    .bar(subset='Porcentaje',vmin=0, vmax=100, color='#4494DA')
estilo_03
```

# Mision 4: Equipo Logistica
> 
> Se debe replantear la estructura y organizacion de provisionamiento de productos, para esto, se solicita la creacion de una visualizacion que permita identificar el estandar de __la cantidad de productos solicitados por mes por departamento__.
---

```python
meses = df_01.copy()
meses = meses.sort_values(by='fecha_pedido', ascending=True)
meses
```

```python
meses['mes'] = meses['fecha_pedido'].dt.strftime('%Y - %b')
meses = meses.reset_index(drop=True)
meses
```

```python
ventas_mes = meses.pivot_table(index=meses['departamento'], columns='mes', values='cantidad', aggfunc='sum', sort=False)
ventas_mes
```

```python
estilo_04 = ventas_mes.style
estilo_04
```

```python
estilo_04.set_sticky(axis='index')
# '''
# Para fijar las columnas
# estilo_app.set_sticky(axis=1)
# estilo_app.set_sticky(axis='columns')
# '''
```

# Mision 5: Informe de Desempeno
> 
> El informe de desempeno es un documento importante para la empresa. En este, todos los que trabajan en la tienda pueden entender como se encuetra el estado actual de la empresa. Debdio a esto. las tablas bien construidas son esenciales para mostrar los datos de la mejor forma.
> 
> Se necesita construir una visualizacion que permita endener ___la relacion entre tipos de clientes y modo de envio de productos de acuerdo con las ventas___. De esta forma, se podra destacar en el informe cual es el modo de envio y cuales son los que mas generaron lucro en las ventas.
---

```python
df_clientes = df_01.pivot_table(index='segmento_cliente', columns='modo_envio', values='ventas', aggfunc='sum', sort=False)
df_clientes
```

```python
df_clientes['Total'] = df_clientes.sum(axis=1).round(2)     # Columna total
df_clientes.loc['Total'] = df_clientes.sum().round(2)       # Fila total
df_clientes
```

```python
estilo_05 = df_clientes.style.format('{:,.2f}')
estilo_05
```

```python
tabla = {
    'selector': 'td,th:not(.index_name)',
    'props': 'text-align: center; font-family: Arial; font-weight: normal; background-color: white'
}

index = {
    'selector': '.index_name',
    'props': 'text-align: right; font-style: italic; font-family: Arial; font-weight: normal; color: #4494DA; background-color: white'
}

estilo_05.set_table_styles([tabla, index], overwrite=False)
```

```python
estilo_05.set_table_styles({
    'Total': [
        {'selector': 'td',
         'props': 'border-top: 2px solid #4494DA'}
        ,{'selector': 'th',
         'props': 'border-top: 2px solid #4494DA'}
    ],
    'B2C': [
        {'selector': 'td',
         'props': 'border-top: 2px solid #4494DA'}
        ,{'selector': 'th',
         'props': 'border-top: 2px solid #4494DA'}
    ]
}, overwrite=False, axis=1)
```

```python
estilo_05.set_table_styles({
    'Total': [
        {'selector': '.true',
         'props': 'background-color: #E5E5E5; color: #4494DA'}
    ]
}, overwrite=False, axis=0)

colores_columna = pd.DataFrame(['false','true','false'], index=df_clientes['Total'].index, columns=['Total'])

estilo_05.set_td_classes(colores_columna)
```

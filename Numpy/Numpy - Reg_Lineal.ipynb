```python
n = np.size(precios_moscu)
y = precios_moscu
x = np.arange(n)
print(x)
```

```python
alpha = (n*np.sum(x*y)-np.sum(x)*np.sum(y)) / (n*np.sum(x**2)-np.sum(x)**2)
beta = np.mean(y)-alpha*np.mean(x)
```

```python
print(f'La pendiente es: {alpha} y el intercepto es: {beta}')
```

```python
y = (alpha*x)+beta
y
```

```python
np.linalg.norm(precios_moscu-y)
```

```python
plt.plot(x, precios_moscu)
plt.plot(x, y)
plt.plot(23,alpha*23+beta,'*r')
plt.plot(68,alpha*68+beta,'*r')
```

```python
while True:
    try:
        peso = float(input("Ingrese su peso en kg: "))
        altura = float(input("Ingrese su altura en metros: "))
        imc = peso / altura ** 2
        print(f"Su IMC es: {imc}")
        if imc < 18.5:
            print("Bajo peso")
        elif imc < 25:
            print("Normal")
        elif imc < 30:
            print("Sobrepeso")
        else:
            print("Obesidad")
        break
    except ValueError:
        print("Ingrese valores numéricos válidos.") 
```

```python
import random

def promedio_3(notas_aleatorias):
    promedio = sum(notas_aleatorias) / len(notas_aleatorias)
    if promedio >= 7:
        situacion = "Aprobado"  
    else:
        situacion = "Reprobado"
    return promedio, situacion  

notas_aleatorias = [random.randint(1, 10) for _ in range(10)]   
print(f"Notas aleatorias: {notas_aleatorias}")

promedio_3(notas_aleatorias)
print(f'El estudiante obtubo un promedio de: {promedio_3(notas_aleatorias)[0]} y su situación es: {promedio_3(notas_aleatorias)[1]} ')
```
